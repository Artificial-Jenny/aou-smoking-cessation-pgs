Following Keller (2014), all covariate × PGS and covariate × treatment interactions are included.
Robustness checks: (i) adding psychiatric comorbidities and their interactions, and (ii) inverse probability weighting (WeightIt).

## 3. Robustness to psychiatric comorbidities (Tables S16–S19)

Refits the base models, adding psychiatric main effects, then psychiatric main effects plus their interactions with treatment and with PGS.

```r
# =========================================================================
# 1. Psychiatric-adjusted moderation models (Keller 2014 consistent)
# 2. Forest plot of all PRS x treatment interaction terms
# =========================================================================

library(dplyr)
library(purrr)
library(tidyr)
library(broom)
library(ggplot2)

EUR <- DF_smokerV3 %>% filter(ancestry == "EUR")
AFR <- DF_smokerV3 %>% filter(ancestry == "AFR")

psych_vars <- c("ADHD", "Depression", "SCZ",
                "Bipolar_Disorder", "PTSD", "Anxiety")

# -------------------------------------------------------------------------
# Formula builder: keeps all four models structurally identical and
# guarantees the correct PRS is used in the key moderation term.
# -------------------------------------------------------------------------
build_formula <- function(prs, psych = c("none", "main", "full")) {

  psych <- match.arg(psych)
  pcs   <- paste0("PC", 1:10)

  base <- c(
    "age", "sex_fac", "treatment_group_factor", prs, pcs,
    # Covariate x Exposure
    "age:treatment_group_factor",
    "sex_fac:treatment_group_factor",
    # Covariate x Genotype
    paste0("age:", prs),
    paste0("sex_fac:", prs),
    paste0(pcs, ":", prs),
    # Key moderation term
    paste0("treatment_group_factor:", prs)
  )

  extra <- switch(
    psych,
    none = NULL,
    main = psych_vars,
    full = c(psych_vars,
             paste0(psych_vars, ":treatment_group_factor"),  # Cov x Exposure
             paste0(psych_vars, ":", prs))                   # Cov x Genotype
  )

  as.formula(paste("quit_success ~", paste(c(base, extra), collapse = " + ")))
}

fit <- function(prs, data, psych) {
  glm(build_formula(prs, psych), data = data, family = "binomial")
}

# -------------------------------------------------------------------------
# Fit all models. Unique, self-documenting names.
# -------------------------------------------------------------------------
models <- list(
  # Base
  "PRS-CES|EUR|Base"        = fit("PRS_CES_scaled", EUR, "none"),
  "PRS-CES|AFR|Base"        = fit("PRS_CES_scaled", AFR, "none"),
  "PRS-CPD|EUR|Base"        = fit("PRS_CPD_scaled", EUR, "none"),
  "PRS-CPD|AFR|Base"        = fit("PRS_CPD_scaled", AFR, "none"),
  # Psychiatric main effects only
  "PRS-CES|EUR|+Psych"      = fit("PRS_CES_scaled", EUR, "main"),
  "PRS-CES|AFR|+Psych"      = fit("PRS_CES_scaled", AFR, "main"),
  "PRS-CPD|EUR|+Psych"      = fit("PRS_CPD_scaled", EUR, "main"),
  "PRS-CPD|AFR|+Psych"      = fit("PRS_CPD_scaled", AFR, "main"),
  # Psychiatric main effects + all covariate interactions (Keller 2014)
  "PRS-CES|EUR|+Psych x"    = fit("PRS_CES_scaled", EUR, "full"),
  "PRS-CES|AFR|+Psych x"    = fit("PRS_CES_scaled", AFR, "full"),
  "PRS-CPD|EUR|+Psych x"    = fit("PRS_CPD_scaled", EUR, "full"),
  "PRS-CPD|AFR|+Psych x"    = fit("PRS_CPD_scaled", AFR, "full")
)

# Sanity check: confirm each model used the intended PRS
map_chr(models, ~ paste(grep("PRS_", names(coef(.x)), value = TRUE)[1]))

# =========================================================================
# Forest plot of the key interaction terms
# =========================================================================

forest_df <- imap_dfr(models, function(m, label) {
  tidy(m, conf.int = TRUE) %>%
    filter(grepl("treatment_group_factor", term),
           grepl("PRS_(CES|CPD)_scaled", term),
           grepl(":", term)) %>%
    mutate(model = label)
}) %>%
  separate(model, into = c("PRS", "Ancestry", "Model"), sep = "\\|") %>%
  mutate(
    Treatment = term %>%
      gsub("treatment_group_factor", "", .) %>%
      gsub(":?PRS_(CES|CPD)_scaled:?", "", .),
    Treatment = factor(Treatment, levels = c("NRT", "Varenicline", "Both")),
    Ancestry  = factor(Ancestry,  levels = c("EUR", "AFR")),
    PRS       = factor(PRS,       levels = c("PRS-CES", "PRS-CPD")),
    Model     = factor(Model,     levels = c("Base", "+Psych", "+Psych x")),
    sig       = p.value < 0.05
  )

forest <- ggplot(forest_df,
                 aes(x = estimate, y = Treatment,
                     colour = Model, shape = sig)) +
  geom_vline(xintercept = 0, linetype = "dashed", colour = "grey40") +
  geom_pointrange(aes(xmin = conf.low, xmax = conf.high),
                  position = position_dodge(width = 0.6),
                  fatten = 2.5) +
  facet_grid(PRS ~ Ancestry) +
  scale_shape_manual(values = c(`FALSE` = 1, `TRUE` = 16),
                     labels = c("p >= 0.05", "p < 0.05"),
                     name = NULL) +
  scale_colour_manual(values = c("#0072B2", "#E69F00", "#009E73")) +
  labs(
    x = expression(paste("PRS ", symbol("\264"),
                         " treatment interaction (log-odds, 95% CI)")),
    y = NULL,
    colour = "Model specification"
  ) +
  theme_classic(base_size = 12) +
  theme(strip.background = element_blank(),
        strip.text = element_text(face = "bold"),
        legend.position = "bottom")

forest

ggsave("FigureS_interaction_forest.pdf", forest, width = 9, height = 6)

# Export the underlying numbers for the supplementary table
write.csv(forest_df, "interaction_estimates.csv", row.names = FALSE)
```

## 4. Inverse probability weighting (Tables S20–S21)

Propensity scores for each active treatment vs. no treatment are estimated from PGS, age, sex, PCs and psychiatric comorbidities. Weights are trimmed at the 99th percentile, balance is checked (|SMD| < 0.1), and the weighted interaction model is refit.

```r
# =========================================================================
# IPW sensitivity analysis: PRS x treatment moderation
# Each active treatment contrasted against No Treatment, by ancestry & PRS
# =========================================================================

library(dplyr)
library(WeightIt)
library(cobalt)
library(survey)
library(purrr)
library(broom)

pcs        <- paste0("PC", 1:10)
psych_vars <- c("ADHD", "Depression", "SCZ",
                "Bipolar_Disorder", "PTSD", "Anxiety")

# -------------------------------------------------------------------------
# `contrast` is the active treatment level compared against "No Treatment"
# -------------------------------------------------------------------------
run_ipw <- function(prs, anc, contrast,
                    data = DF_smokerV3, trim_at = 0.99,
                    estimand = "ATE") {

  d <- data %>%
    filter(ancestry == anc,
           treatment_group_factor %in% c("No Treatment", contrast)) %>%
    droplevels() %>%
    mutate(treatment_group_factor =
             relevel(treatment_group_factor, ref = "No Treatment"))

  vars_needed <- c("quit_success", "treatment_group_factor", prs,
                   "age", "sex_fac", pcs, psych_vars)
  n_before <- nrow(d)
  d <- d %>% filter(if_all(all_of(vars_needed), ~ !is.na(.)))

  n_treated <- sum(d$treatment_group_factor == contrast)
  message(sprintf("%s | %s | %s vs No Treatment: N = %d (%d treated), %d dropped",
                  prs, anc, contrast, nrow(d), n_treated, n_before - nrow(d)))

  # --- propensity score weights ----------------------------------------
  ps_formula <- as.formula(paste(
    "treatment_group_factor ~",
    paste(c(prs, "age", "sex_fac", pcs, psych_vars), collapse = " + ")
  ))

  w <- weightit(ps_formula, data = d, method = "ps", estimand = estimand)
  w <- trim(w, at = trim_at)

  stopifnot(length(w$weights) == nrow(d))

  bal <- bal.tab(w, thresholds = c(m = 0.1), un = TRUE)

  # --- weighted outcome model ------------------------------------------
  out_formula <- as.formula(paste(
    "quit_success ~ treatment_group_factor *", prs, "+",
    "age + sex_fac +",
    "age:treatment_group_factor + sex_fac:treatment_group_factor +",
    paste0("age:", prs, " + sex_fac:", prs, " +"),
    paste(pcs, collapse = " + "), "+",
    paste(paste0(pcs, ":", prs), collapse = " + ")
  ))

  des <- svydesign(ids = ~1, weights = w$weights, data = d)
  fit <- svyglm(out_formula, design = des, family = quasibinomial())

  list(data = d, weights = w, balance = bal, model = fit,
       n = nrow(d), n_treated = n_treated,
       ess = sum(w$weights)^2 / sum(w$weights^2))
}

# -------------------------------------------------------------------------
# Run every combination: 2 PRS x 2 ancestries x 3 contrasts
# -------------------------------------------------------------------------
grid <- expand.grid(
  prs      = c("PRS_CES_scaled", "PRS_CPD_scaled"),
  anc      = c("EUR", "AFR"),
  contrast = c("Varenicline", "Both", "NRT"),
  stringsAsFactors = FALSE
)

ipw <- pmap(grid, function(prs, anc, contrast) {
  tryCatch(run_ipw(prs, anc, contrast),
           error = function(e) {
             warning(sprintf("FAILED: %s | %s | %s -- %s",
                             prs, anc, contrast, conditionMessage(e)))
             NULL
           })
})
names(ipw) <- with(grid, paste(gsub("_scaled", "", prs), anc, contrast,
                               sep = "|"))
ipw <- compact(ipw)   # drop any that failed to converge





```
