# Supplement · Do PGS predict who receives treatment?

Paper section: **Supplementary Results** (treatment allocation; motivates the IPW analysis in Aim 2, and is cited in the Discussion).

Logistic regressions of receiving any cessation treatment on each PGS, with the same nested covariate sets as Aim 1,
plus PGS distributions by treatment group.

**Inputs:** `DF_smokerV2.csv`, `DF_smokerV3.csv`

```r
# Setup: run from the repository root inside the All of Us Researcher Workbench
source("R/helpers.R")
install_if_missing(c("tidyverse", "ggdist", "ggsignif"))
```

```r
DF_smokerV2 <- load_from_bucket("DF_smokerV2.csv") %>% prep_analysis_factors()
DF_smokerV3 <- load_from_bucket("DF_smokerV3.csv") %>% prep_analysis_factors()
```

## PGS + demographics

```r
library(dplyr)
model_1 <- glm(has_treatment ~ age + sex_fac + PRS_CPD_scaled +
                   PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10, 
               data = DF_smokerV2%>%filter(ancestry == "EUR"), family = "binomial")
summary(model_1)$coefficients
nobs(model_1)

model_1_afr <- glm(has_treatment ~ age + sex_fac + PRS_CPD_scaled +
                   PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10, 
               data = DF_smokerV2%>%filter(ancestry == "AFR"), family = "binomial")
summary(model_1_afr)$coefficients
nobs(model_1_afr)

model_2 <- glm(has_treatment ~ age + sex_fac + PRS_CES_scaled +
                   PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10, 
               data = DF_smokerV2%>%filter(ancestry == "EUR"), family = "binomial")
summary(model_2)$coefficients
nobs(model_2)

model_2_afr <- glm(has_treatment ~ age + sex_fac + PRS_CES_scaled +
                   PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10, 
               data = DF_smokerV2%>%filter(ancestry == "AFR"), family = "binomial")
summary(model_2_afr)$coefficients
nobs(model_2_afr)
```

## PGS distribution by treatment group

```r
library(ggdist)
library(ggsignif)
library(tidyverse)

make_prs_plot <- function(anc) {
  plot_data <- DF_smokerV3 %>%
    filter(ancestry == anc) %>%
    mutate(
      group = case_when(
        is.na(treatment_group)                    ~ "General Smoking\nPopulation",
        treatment_group == "Varenicline Only"       ~ "Varenicline",
        treatment_group == "NRT Only"               ~ "NRT",
        treatment_group == "Both NRT + Varenicline" ~ "NRT + Varenicline",
        TRUE ~ NA_character_
      ),
      group = factor(group, levels = c("General Smoking\nPopulation", "NRT",
                                       "Varenicline", "NRT + Varenicline"))
    ) %>%
    filter(!is.na(group), !is.na(PRS_CPD_scaled))

  print(table(plot_data$group))   # check group sizes

  ggplot(plot_data, aes(x = group, y = PRS_CPD_scaled, fill = group)) +
    stat_slab(side = "right", position = position_nudge(x = 0.15),
              adjust = 1.5, scale = 0.6, alpha = 0.8) +
    geom_point(
      data = plot_data %>% group_by(group) %>% sample_frac(0.3),
      position = position_jitter(width = 0.08, height = 0, seed = 123),
      size = 0.5, alpha = 0.3, color = "black"
    ) +
    geom_boxplot(width = 0.15, outlier.shape = NA, alpha = 0.8,
                 position = position_nudge(x = -0.05)) +
    scale_fill_manual(values = c(
      "General Smoking\nPopulation" = "gray70",
      "NRT"               = "#E69F00",
      "Varenicline"       = "#CC79A7",
      "NRT + Varenicline" = "#0072B2"
    )) +
    geom_signif(
      comparisons = list(
        c("General Smoking\nPopulation", "NRT"),
        c("General Smoking\nPopulation", "Varenicline"),
        c("General Smoking\nPopulation", "NRT + Varenicline")
      ),
      test = "wilcox.test", map_signif_level = TRUE,
      step_increase = 0.1, textsize = 5
    ) +
    labs(title = "PRS-CPD Distribution by Treatment Group",
         subtitle = paste(anc, "Ancestry"),
         x = "", y = "PRS-CPD (Standardized)") +
    theme_minimal(base_size = 20) +
    theme(
      plot.title = element_text(face = "bold", size = 24, hjust = 0.5),
      plot.subtitle = element_text(size = 20, hjust = 0.5),
      axis.title.y = element_text(face = "bold", size = 22),
      axis.text = element_text(size = 20),
      axis.line = element_line(color = "black", linewidth = 0.5),
      axis.ticks = element_line(linewidth = 0.5),
      panel.grid.minor = element_blank(),
      panel.grid.major.x = element_blank(),
      panel.grid.major.y = element_line(color = "gray90"),
      legend.position = "none"
    )
}

p1 <- make_prs_plot("EUR")
p2 <- make_prs_plot("AFR")
print(p1)
print(p2)
```
