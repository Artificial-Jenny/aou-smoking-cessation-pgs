# 05 · Aim 1: PGS associations with smoking cessation


Logistic regressions of quit success on CPD<sup>PGS</sup> or CES<sup>PGS</sup>, run separately by ancestry:

| Model | Covariates |
|---|---|
| 1 | PGS + age + sex + 10 PCs |
| 2 | Model 1 + psychiatric comorbidities |
| 3 | Model 1 + SDoH |
| 4 (primary) | Model 1 + psychiatric + SDoH |

**Input:** `DF_smokerV3.csv`

```r
DF_smokerV3 <- load_from_bucket("DF_smokerV3.csv") %>% prep_analysis_factors()
```

## Model 1: PGS + demographics

```r
model1_CPD_EUR <- glm(quit_success ~ age + sex_fac + PRS_CPD_scaled +
                 PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10,
               data = DF_smokerV3 %>% filter(ancestry == "EUR"), family = "binomial")
summary(model1_CPD_EUR)$coefficients
nobs(model1_CPD_EUR)

model1_CPD_AFR <- glm(quit_success ~ age + sex_fac + PRS_CPD_scaled +
                 PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10,
               data = DF_smokerV3 %>% filter(ancestry == "AFR"), family = "binomial")
summary(model1_CPD_AFR)$coefficients
nobs(model1_CPD_AFR)
```

```r
model1_CES_EUR <- glm(quit_success ~ age + sex_fac + PRS_CES_scaled +
                 PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10,
               data = DF_smokerV3 %>% filter(ancestry == "EUR"), family = "binomial")
summary(model1_CES_EUR)$coefficients
nobs(model1_CES_EUR)

model1_CES_AFR <- glm(quit_success ~ age + sex_fac + PRS_CES_scaled +
                 PC1 + PC2 + PC3 + PC4 + PC5 + PC6 + PC7 + PC8 + PC9 + PC10,
               data = DF_smokerV3 %>% filter(ancestry == "AFR"), family = "binomial")
summary(model1_CES_AFR)$coefficients
nobs(model1_CES_AFR)
```
